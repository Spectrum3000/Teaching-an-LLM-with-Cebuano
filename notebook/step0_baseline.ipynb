# Step 0 - Baseline generation quality

Member A task: run untouched `distilgpt2` on Cebuano prompts and preserve the exact before-training outputs.

```python
from pathlib import Path
from transformers import pipeline

PROJECT = next(path for path in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent] if (path / 'carl work').exists() or (path / 'data').exists())
CARL = PROJECT / 'carl work' if (PROJECT / 'carl work').exists() else PROJECT
OUTPUTS = CARL / 'outputs'
OUTPUTS.mkdir(parents=True, exist_ok=True)
print('Carl work:', CARL)
print('Outputs:', OUTPUTS)
```

```python
generator = pipeline('text-generation', model='distilgpt2', device=-1)
prompts = [
    'Ang mga estudyante sa Cebu',
    'Nagsugod na ang',
    'Ang kinabuhi sa',
    'Miadto siya sa',
    'Sa matag adlaw,'
]
records = []
for prompt in prompts:
    result = generator(prompt, max_new_tokens=50, do_sample=False, pad_token_id=generator.tokenizer.eos_token_id)
    text = result[0]['generated_text']
    records.append((prompt, text))
    print(f'PROMPT: {prompt}')
    print(f'OUTPUT: {text}')
    print('-' * 80)

with (OUTPUTS / 'step0_generations.txt').open('w', encoding='utf-8') as file:
    for prompt, text in records:
        file.write(f'PROMPT: {prompt}\nOUTPUT: {text}\n' + '-' * 80 + '\n')
print('Saved:', OUTPUTS / 'step0_generations.txt')
```
