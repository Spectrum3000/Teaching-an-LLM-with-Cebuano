```python
# Check for very short entries
short_entries = [i for i, ex in enumerate(read_dataset) if len(ex['text']) < 20]
print(f"Very short entries: {len(short_entries)}")

# Check for possibly English entries
def likely_english(text):
    english_words = ['the', 'is', 'are', 'was', 'were', 'this', 'that']
    words = text.lower().split()
    matches = sum(1 for w in words if w in english_words)
    return matches > 2

english_entries = [i for i, ex in enumerate(read_dataset) if likely_english(ex['text'])]
print(f"Possibly English entries: {len(english_entries)}")
```

```python
from datasets import load_dataset

ner_train = load_dataset(
    "josephimperial/CebuaNER",
    split="train"
)

sentences = []
current_tokens = []

for row in ner_train["text"]:
    row = row.strip()

    # Document or sentence boundary
    if not row or row.startswith("-DOCSTART-"):
        if current_tokens:
            sentences.append(" ".join(current_tokens))
            current_tokens = []
        continue

    parts = row.split()

    if parts:
        current_tokens.append(parts[0])

if current_tokens:
    sentences.append(" ".join(current_tokens))

ner_texts = [
    sentence for sentence in sentences
    if len(sentence) > 20
]

print(f"CebuaNER: {len(ner_texts)} sentences reconstructed")

print("\nSample:")
for sentence in ner_texts[:5]:
    print(f"  {sentence}")
```

```python
import random 

read_dataset = load_dataset(
    "filbench/cebuano-readability",
    split="test"
)

read_texts = [ex['text'].strip() for ex in read_dataset
              if len(ex['text'].strip()) > 20]

print(f"Readability: {len(read_texts)} passages")
print(f"CebuaNER: {len(ner_texts)} sentences")
print(f"Total: {len(ner_texts) + len(read_texts)} entries")

all_texts = read_texts + ner_texts

random.seed(42)
random.shuffle(all_texts)

# Write to the data/ folder in your repo
output_path = "../data/cebuano_corpus.txt"
with open(output_path, "w", encoding="utf-8") as f:
    for text in all_texts:
        f.write(text.strip() + "\n\n")

print(f"Corpus written to {output_path}")
```

```python
with open("../data/cebuano_corpus.txt", "r", encoding="utf-8") as f:
    entries = f.read().split("\n\n")
    entries = [e.strip() for e in entries if e.strip()]

sample = random.sample(entries, 20)
print(f"Total entries: {len(entries)}")
print("\nRandom sample of 20:")
for i, entry in enumerate(sample):
    print(f"\n[{i+1}] {entry[:150]}{'...' if len(entry) > 150 else ''}")
```
