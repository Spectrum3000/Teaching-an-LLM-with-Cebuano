# Data Preparation Reference

This reproduces the important checks from the historical `file_data_prep.ipynb.ipynb` notebook. It uses the actual Hub splits and explains the CoNLL-style CebuaNER conversion used by the Member A training notebooks.

```python
from datasets import load_dataset

read_dataset = load_dataset('filbench/cebuano-readability', split='test')
ner_dataset = load_dataset('josephimperial/CebuaNER', split='train')
print('Readability:', read_dataset)
print('Readability columns:', read_dataset.column_names)
print('CebuaNER:', ner_dataset)
print('CebuaNER columns:', ner_dataset.column_names)
print('CebuaNER first row:', ner_dataset[0])
```

```python
def reconstruct_cebuano_documents(rows):
    documents = []
    current_tokens = []
    for raw_row in rows:
        row = raw_row.strip()
        if not row or row.startswith('-DOCSTART-'):
            if current_tokens:
                documents.append(' '.join(current_tokens))
                current_tokens = []
            continue
        current_tokens.append(row.split()[0])
    if current_tokens:
        documents.append(' '.join(current_tokens))
    return [document for document in documents if len(document) > 20]

ner_texts = reconstruct_cebuano_documents(ner_dataset['text'])
read_texts = [row['text'].strip() for row in read_dataset if len(row['text'].strip()) > 20]
print('CebuaNER reconstructed units:', len(ner_texts))
print('Readability passages:', len(read_texts))
print('Combined entries:', len(ner_texts) + len(read_texts))
print('Sample:', ner_texts[0])
```

The historical notebook calls these reconstructed units sentences, although the raw CebuaNER file uses blank lines as boundaries. The training corpus already generated in the repository is the shared input for Member A; this reference cell is for schema verification and reproducibility.
