# Step 0 - Baseline generation quality

Member A responsibility: run untouched `distilgpt2` on Cebuano prompts, preserve the exact outputs, and use them as the before-training evidence.

```python
from pathlib import Path
from transformers import pipeline

candidates = [Path.cwd() / 'member A', Path.cwd().parent, Path.cwd().parent.parent / 'member A']
MEMBER_A = next(path for path in candidates if (path / 'outputs').exists())
OUTPUTS = MEMBER_A / 'outputs'
OUTPUTS.mkdir(parents=True, exist_ok=True)
print('Member A folder:', MEMBER_A)
print('Outputs folder:', OUTPUTS)
```

```python
generator = pipeline('text-generation', model='distilgpt2', device=-1)
prompts = [
    'Ang mga estudyante sa Cebu',
    'Nagsugod na ang',
    'Ang kinabuhi sa',
    'Miadto siya sa',
    'Sa matag adlaw,'
]
records = []
for prompt in prompts:
    result = generator(prompt, max_new_tokens=50, do_sample=False, pad_token_id=generator.tokenizer.eos_token_id)
    text = result[0]['generated_text']
    records.append((prompt, text))
    print(f'PROMPT: {prompt}')
    print(f'OUTPUT: {text}')
    print('-' * 80)

with (OUTPUTS / 'step0_generations.txt').open('w', encoding='utf-8') as file:
    for prompt, text in records:
        file.write(f'PROMPT: {prompt}\nOUTPUT: {text}\n' + '-' * 80 + '\n')
print('Saved:', OUTPUTS / 'step0_generations.txt')
```
